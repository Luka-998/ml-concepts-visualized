### Principal Component Analysis

#### __1) Dimensionality reduction__

####  When faced with a large set of corre- lated variables, principal components allow us to summarize this set with a smaller number of representative variables that collectively explain most of the variability in the original 

#### PCA directions are represented as directions in the **feature space** along which the original data are __highly variable__

### __2) Visualization__

#### If we have a data __n x p__ where n= number of observations and p = number of features, in EDA, we would like to visualize scatterplot.

#### If number of feature is 10 , and we examine 2D scatterplot of the data , we will end up with (p /2) = p * (p-1)/2 => 45 scatterplots!

#### if __p is large__, we will not be able to look at all of them, and most of them will not be informative because each one of them contain just a small fraction of total infromation present in the data set.

#### Considering this case, we would like to find a low-dimensional representation of data that captures as much information as possible.

#### __INTUITION__: 

- each of n observations lives in a p dimensional space, but not all dimensions are intersting.

- PCA seeks small number of dimensions, where the "interesting" part is measured by the amount that the observations vary along each dimension

- Each dimension of PCA is linear combination of _p_ features

### Making the covariance matrix S 

- covariance is the measure of how two variables correlate to eachother

- positive score indicates that variables are correlated, meaning that they move in the same direction

- my matrix is shape of (5,10). To make covariance matrix i will use the S:

### S  = 1 / n - 1 * np.dot(Xc,Xc.T)

- 1 / n - 1 because we derive covariance from the sample of population

- Xc means that our sample is centered which i will do manually 

we can not multiply the native shape 5,10 x 5,10 because inner dimmensions need to match

if we have 5 features, covariance matrix will be 5x5 matrix:
