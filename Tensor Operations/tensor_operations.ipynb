### **Mathematical understanding of tensor operations**

#### 1) **Element wise operations**

Tensor -> it's higher dimension vector such as 2D >

Matrix with m,n can be interpreted as 2d -tensor

```python
import numpy as np
```

##### Considering one of the activation functions such as __ReLU__:

```python
x = np.arange(10).reshape(5,2)
y = np.arange(10).reshape(5,2)
def naive_element_wise_addition(x,y):
    assert len(x.shape) == 2
    assert x.shape == y.shape
  
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            x[i,j] +=y[i,j]
    return x
```

```python
print(f"X:\n{x}\n{'*'*10}\nY:\n{y}")
```

```python
z = naive_element_wise_addition(x,y)
print(z)
```

#### 3) **Tensor dot @**

Most useful tensor operation ( not to be confused with element wise dot product such as skalar (1,) @ vector(4,)  for example.)

```python
x = np.array([1,2,3])
y = np.array([3,4,5])
```

```python
def naive_vector_dot(x,y):
    assert len(x.shape) == 1
    assert len(y.shape) == 1
    assert x.shape[0] == y.shape[0]

    z = 0
    for i in range(x.shape[0]):
        z += x[i] * y[i]
    return z
```

```python
z = naive_vector_dot(x,y)
print(z)
```

The Dot product between __vectors__ is a SCALAR

Only vectors with same number of elements are compatible for the Dot product!

#### 3.1) **Dot product between matrix and vector**:

```python
x = np.arange(6).reshape(2,3)
y = np.arange(3).reshape(3,)
```

```python
def naive_matrix_vector_dot(x,y):
    assert len(x.shape) == 2
    assert len(y.shape) == 1

    assert x.shape[1] == y.shape[0]

    z = np.zeros_like(x.shape)

    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            z += x[i,j] * y[i]
    return z
```

```python
z = naive_matrix_vector_dot(x,y)
print(z)
```

#### 3.2) **Dot product between 2 matrix**:

We can take the dot product between 2 matrices only and only if x.shape[1] == y.shape[0]

i.e only if column of first matrix match the row of second matrix
