### **Numpy Model (Forward Prop in NumPy)**

#### 2 Layers (sigmoid activation)
#### Input vector will consist of 1 example such as a [0] = [a] [ value for x1, value for x2 ]

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.layers import Dense
from tensorflow.keras import layers
from tensorflow.keras import Sequential
```

```python
dataset = pd.read_csv(file_path)
dataset.head()
```

```python
X = dataset.iloc[:,0:2]
y = dataset.iloc[:,-1]
```

```python
def sigmoid(z):
    return 1/1+np.exp(z)
```

```python
def Dense(a_in,W,b):
    # network architecture is not predefined, it adjust to shapes
    units = W.shape[1]
    a_out = np.zeros(units)
    for i in range(units):
        w = W[:,i]                                    
        z = np.dot(w, a_in) + b[i]         
        a_out[i] = sigmoid(z)               
    return(a_out)
```

```python
def my_sequential(x,W1,b1,W2,b2):
    a1 = Dense(x,W1,b1)
    a2 = Dense(a1,W2,b2)
    return a2
```

```python
W1_tmp = np.array( [[-8.93,  0.29, 12.9 ], [-0.1,  -7.32, 10.81]] )
b1_tmp = np.array( [-9.82, -9.28,  0.96] )
W2_tmp = np.array( [[-31.18], [-27.59], [-32.56]] )
b2_tmp = np.array( [15.41] )
```

```python
def predict(X,W1,b1,W2,b2):
    # prediction goes - > column wise shape[0]
    m,n = X.shape
    p = np.zeros((m,n))
    for j in range(m): # iterate through rows of Input
        p[j,0] = my_sequential(X[j],W1,b1,W2,b2)
    return p
```

```python
X_tst = np.array([
    [200,13.9],  # postive example
    [200,17]])   # negative example

# data normalization
norm_l = tf.keras.layers.Normalization(axis=-1)
X_tstn = norm_l(X_tst)  # remember to normalize
predictions = predict(X_tstn, W1_tmp, b1_tmp, W2_tmp, b2_tmp)
```

```python
y_pred = (predictions>=0.5).astype(int)
y_pred
```
