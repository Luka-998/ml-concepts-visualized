```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = np.linspace(-5,5,1000)
y = np.linspace(-5,5,1000)
xv,yv = np.meshgrid(x,y)
z = np.sqrt(xv**2+yv**2) >= 1
```

```python
z.astype(float)
```

```python
sns.heatmap(data=z,cmap='rocket')
plt.show()
plt.savefig('Dot.jpg')
```
