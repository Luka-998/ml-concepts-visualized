### **Neural Network from scratch using only NumPy**

This project implements a simple fully-connected neural network from scratch using only NumPy. The goal is to understand the core mechanics of forward propagation, parameter initialization, and data flow through a network without relying on high-level frameworks.

#### **Architecture**

The model is a 2-layer feedforward neural network:

Input layer: 2 features (testing phase)

Hidden layer: 4 neurons (sigmoid activation)

Output layer: 1 neuron (sigmoid activation)

#### **Notes**


- Testing phase - input vector will be synthetic batch sample of few values for a1, a2
- Parameters are initialized with small random values
- Output is in range (0, 1), suitable for binary tasks
- Emphasis is on clarity and mathematical correctness
- Production phase - Using real kaggle dataset

```python
import numpy as np
import pandas as pd
import pathlib
```

```python
# Initialization of the sample values for input (a_in):
X = np.array([150, 15]) 

print(X.shape)
```

```python
# define the activation function
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def get_shape(x):
    return x.shape
```

```python
def Dense(units,input_dim):
    W = np.random.rand(units,input_dim) #Ovde mora da bude ceo broj za units 
    b = np.random.rand(units,) 
    activation = sigmoid
    layer = {
        'W':W,
        'b':b,
        'activation':activation
    }
    return layer


```

```python
layer = Dense(4,2)
```

##### **Feed Forward propagation between initial input __a__\_[0] towards the output a_[2]**

Inspired by tf.keras.Sequential used for stacking layers sequentially

Information flow is in order from first layer to the last one.

```python
X = np.array([14,15])
```

```python
def Sequential(num_layers,X):
    # Dynamic making layers stacking
    units_per_layer = [4,1] # Neuros in Layer_1, Layer_n-1
    if num_layers != len(units_per_layer):
        print(f"Layer number is not same as the number of neuron /per layer!\nLayers: {num_layers}\nNeurons in each layer:{units_per_layer}")
    
    layers_total = []

    for i in range(num_layers):
        if i == 0: # input layer shape
            input_dim = X.shape
            A = input_dim[0]
        else:
            A = units_per_layer[i-1]
        layer = Dense(units_per_layer[i],A)
        layers_total.append(layer)
    return layers_total
        
```

##### vector X is a feature input from L_0

- After going into L_1 (first hidden layer) we calculate for each unit:

z1 = dot( w11,a_in ) + b11

z2 = dot ( w12,a_in ) + b12

z3 ...

z4 ....

Which means the output vector from L1 is of shape (4,) - > it is the input vector of another layer !


```python
total_layers = Sequential(2,X)
print(total_layers)
```

**Feedforward propagation**: 

Idea behind the feedforward is to pas the list of vectors W and b. For each layer, different W and b is used to calculate a_in[i].

```python
def feedforward(a_in):
    s = Sequential(2,a_in)
    """
    Testing:

    print(s[0]['W'].shape)
    in_shape = a_in.shape
    print(in_shape) # (2,)
    x = sigmoid(np.dot(s[0]['W'],a_in) + s[0]['b'])
    print(x)
    y = sigmoid(np.dot(s[1]['W'],x) + s[1]['b'])
    print(y)
    
    end 
    """

    for i in range(len(s)): # iterate over the layers
        layer = s[i]['activation'](np.dot(s[i]['W'],a_in)+s[i]['b'])
        a_in = layer
        

    return layer

a_out = feedforward(X)
```
